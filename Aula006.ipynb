```python
import pandas as pd
pd.set_option('display.max_columns', None)


filenome = 'clima.csv'
dados = pd.read_csv(filenome, sep=',')
```

```python
dados.head()
```

```python
import matplotlib.pyplot as plt
```

```python
dados.dtypes
```

```python
dados['Data'] = pd.to_datetime(dados['Data'])
dados = dados.dropna(subset=['MinTemp', 'MaxTemp', 'Chuva', 'VelocidadeVento'])
```

```python
dados.dtypes
```

```python
plt.figure(figsize=(12,5))

plt.plot(dados['Data'], dados['MinTemp'], Label='Temp.Mínima')
plt.plot(dados['Data'], dados['MaxTemp'], Label='Temp.Mínima')

plt.title('Temperatura minima e max ao longo do tempo')

plt.xlabel('Data')

plt.ylabel('Temperatura')

plt.legend()

plt.grid()
```

```python
# Scatter - Temperatura x Evaporaçã
plt.figure(figsize=(8,5))

plt.scatter(dados['Umidade9'], dados['Evaporacao'])

plt.title('Titulo')

```

```python
# Barras 

localizacoes = dados['Localizacoes'].value_counts().head(10)
plt.figure(figsize=(12,5))

plt.bar(localizacoes.index, localizacoes.values)

plt.title('Top 1- localizacoes que mais apareceram')

plt.xlabel('Localização')

plt.ylabel('Quantidade')
```
