```python
import pandas as pd
import numpy as np

df = pd.read_csv('clima.csv')

print("--- 1-3. Temperatura ---")
# 1. Dias mais frios têm menor chance de chover amanhã?
# Vamos ordenar por MinTemp e pegar os 10% mais frios vs resto
frios = df.sort_values(by='MinTemp').head(int(len(df)*0.1))
print("ChuvaAmanha nos 10% dias mais frios:\n", frios['ChuvaAmanha'].value_counts(normalize=True))
print("ChuvaAmanha geral:\n", df['ChuvaAmanha'].value_counts(normalize=True))

# 2. Dias mais quentes (MaxTemp) aumentam a chance?
quentes = df.sort_values(by='MaxTemp', ascending=False).head(int(len(df)*0.1))
print("ChuvaAmanha nos 10% dias mais quentes:\n", quentes['ChuvaAmanha'].value_counts(normalize=True))

# 4. ChuvaHoje afeta ChuvaAmanha?
print("\n--- 4-6. Chuva ---")
print(pd.crosstab(df['ChuvaHoje'], df['ChuvaAmanha'], normalize='index'))

# 6. Frequência de dias com chuva vs sem chuva
print("\nFrequência ChuvaHoje:\n", df['ChuvaHoje'].value_counts(normalize=True))

# 7. DirecaoVento
print("\n--- 7-10. Vento ---")
print(pd.crosstab(df['DirecaoVento'], df['ChuvaAmanha'], normalize='index').sort_values(by='Yes', ascending=False).head(3))

# 11-13. Umidade
print("\n--- 11-13. Umidade ---")
df['Umidade15_bin'] = pd.cut(df['Umidade15'], bins=5)
print(df.groupby('Umidade15_bin')['ChuvaAmanha'].value_counts(normalize=True).unstack())

df['Umidade9_bin'] = pd.cut(df['Umidade9'], bins=5)
print(df.groupby('Umidade9_bin')['ChuvaAmanha'].value_counts(normalize=True).unstack())

# 18-19. Localizacao
print("\n--- 18-19. Localizacao ---")
locais = pd.crosstab(df['Localizacao'], df['ChuvaAmanha'], normalize='index')
print("Locais com mais chuva amanhã (Top 3):\n", locais.sort_values(by='Yes', ascending=False).head(3))

# 20. Alta umidade + muitas nuvens
print("\n--- 20. Umidade + Nuvens ---")
alta_umid = df['Umidade15'] > 80
muitas_nuv = df['Nuvens15'] > 6
print("Prob. chuva com alta umidade e nuvens:\n", df[alta_umid & muitas_nuv]['ChuvaAmanha'].value_counts(normalize=True))
```
