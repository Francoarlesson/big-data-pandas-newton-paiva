## 1 Criar o Box-Plot considerando ChuvaHoje('Sim' ou 'Não')

```python
df_grafico = dados.copy()

sns.boxplot(
    x='ChuvaHoje',
    y='MinTemp',
    data=df_grafico,
    hue='ChuvaHoje'
)
plt.title('Box-Plot - Min-Temp')
plt.ylabel('Temperatura Minima')
plt.show()
```

```python
grupoNao = dados[dados['ChuvaHoje'] == 'No']

Q1 = grupoNao['Umidade9'].quantile(0.25)
Q2 = grupoNao['Umidade9'].quantile(0.50)
Q3 = grupoNao['Umidade9'].quantile(0.75)

IQR = Q3 - Q1

print(f'Q1: {Q1:.2f}')
print(f'Q2 ou Mediana: {Q2:.2f}')
print(f'Q3: {Q3:.2f}')
print(f'IQR: {IQR:.2f}')
```

## 2 - Valor de (Q1, Q2, Q3, IQR, Whisker Superior e Whisker Inferior)

```python
print(f'Q1: {Q1:.2f}') # Q1
print(f'Q2 ou Mediana: {Q2:.2f}') # Q2
print(f'Q3: {Q3:.2f}') # Q3
print(f'IQR: {IQR:.2f}') # IQR

# Whisker Superior
WhiskerSup = Q3 + 1.5 * IQR
print(f'Whisker Superior {WhiskerSup}')

# Whisker Inferior
WhiskerInfe = Q1 - 1.5 * IQR
print(f'Whisker Inferior {WhiskerInfe}')
```
