### 1. Que tipo de dado essas colunas representam?  
Essas colunas representam dados categóricos (também chamados de qualitativos).  Diferente de dados numéricos contínuos (como a temperatura exata em graus Celsius ou a quantidade de chuva em milímetros), os dados categóricos dividem as informações em grupos ou "categorias".Como essas colunas específicas indicam apenas se um evento aconteceu ou não, podemos ser ainda mais precisos e chamá-las de variáveis categóricas binárias (ou booleanas).

## 2. Quais são as categorias possíveis? 
As colunas respondem à pergunta se houve chuva no dia de hoje e no dia de amanhã, as categorias possíveis armazenadas nessas colunas são simplesmente "Yes" e "No".
