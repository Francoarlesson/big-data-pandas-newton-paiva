```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder.getOrCreate()

# Lê tudo como STRING
df = spark.read.csv(
    "clima.csv",
    header=True,
    inferSchema=False
)

# Substitui "Na" por NULL
df = df.replace("Na", None)

# Converte a coluna manualmente
df = df.withColumn(
    "MaxTemp",
    col("MaxTemp").cast("double")
)
df.printSchema()
df.show()
```

```python
# Converte a coluna manualmente 
df = df.withColumns("MinTemp", col("MinTemp").cast("double"))
df.filter((df.MaxTemp > 30)&(df.MinTemp > 20)).select(["Localizacao", "MaxTemp", "MinTemp"]).show()
```
