```python
frequenciaChuvaHoje = df["ChuvaHoje"].value_counts()
print("A frequência de Chuva HOje: \n", frequenciaChuvaHoje)
```

```python
frequenciaChuvaAmanha = df["ChuvaAmanha"].value_counts()
print("A frequência de Chuva Amanhã: \n", frequenciaChuvaAmanha)
```
