```python
import pandas as pd
df = pd.read_csv('clima.csv')
```

## Estatística de Categóricas
## Exercício – variável ‘ChuvaAmanha’:
## Perguntas:

```python
# 1. Em quantos dias choveu no dia seguinte?

diasChuvosos = df['ChuvaAmanha'].value_counts()['Yes']
print(f"Os dias que choveram no dia seguinte: {diasChuvosos}")
```

```python
# 2. Qual categoria é mais comum?

caegoriaComum = df['ChuvaAmanha'].mode()[0]
print(f" As caregoria mais comuns: {caegoriaComum}")
```

```python
# 3. Aproximadamente qual percentual dos dias tem chuva no dia seguinte?
percentuais = df['ChuvaAmanha'].value_counts(normalize=True) * 100
percentual_chuva = percentuais['Yes']
print(f"Percentual de dias com chuva: {percentual_chuva:.2f}%")
```

```python
# 4. Crie um gráfico de barras.

import matplotlib.pyplot as plt

# Gera o gráfico de barras
df['ChuvaAmanha'].value_counts().plot(kind='bar', color=['skyblue', 'lightgray'])

# Adiciona títulos e rótulos para melhor visualização
plt.title('Distribuição da Variável ChuvaAmanha')
plt.xlabel('Choveu no dia seguinte?')
plt.ylabel('Frequência (Quantidade de Dias)')
plt.xticks(rotation=0) # Mantém o texto do eixo X na horizontal
plt.show()
```
