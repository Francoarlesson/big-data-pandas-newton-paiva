```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

# Inicializa a sessão do Spark
spark = SparkSession.builder.appName("Exercicio1_Clima").getOrCreate()

# Lê o arquivo CSV. 
# O parâmetro "nullValue" já cumpre a Etapa 1: Substitui os "NA" do texto original por NULL no Spark.
df_spark = spark.read.option("header", "true") \
                     .option("nullValue", "NA") \
                     .csv("clima.csv")

# Etapas 2 e 3: Converter as colunas Umidade9 e Temperatura9 para o tipo double
df_tratado = df_spark.withColumn("Umidade9", col("Umidade9").cast("double")) \
                     .withColumn("Temperatura9", col("Temperatura9").cast("double"))

# Etapa 4: Criar a nova coluna IndiceClimatico: Temperatura9 - (Umidade9 / 10)
df_tratado = df_tratado.withColumn("IndiceClimatico", col("Temperatura9") - (col("Umidade9") / 10))

# Etapa 6: Filtrar os registros onde Temperatura9 > 25 e Umidade9 > 60
df_filtrado = df_tratado.filter((col("Temperatura9") > 25) & (col("Umidade9") > 60))

# Etapa 5: Selecionar (mostrar) apenas as colunas específicas
df_final = df_filtrado.select("Localizacao", "Temperatura9", "Umidade9", "IndiceClimatico")

# Exibe o resultado final na tela
df_final.show()
```
