```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('clima.csv')
```

## 1. Criar o Box-Plot

```python
plt.figure(figsize=(8, 6))
# Cria o boxplot da Umidade9 separando por ChuvaHoje
sns.boxplot(x='ChuvaHoje', y='Umidade9', data=df)
plt.title('Box-Plot: Umidade às 9h por Ocorrência de Chuva Hoje')
plt.show()
```

### 2. Valores de Q1, Q2, Q3, IQR e Whiskers
### Calculando as métricas para a coluna 'Umidade9' geral. 
### basta aplicar um filtro no df antes de calcular).

```python
Q1 = df['Umidade9'].quantile(0.25)
Q2 = df['Umidade9'].median() # Equivalente a quantile(0.50)
Q3 = df['Umidade9'].quantile(0.75)
IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print("--- MÉTRICAS GERAIS ---")
print(f"Q1 (1º Quartil): {Q1}")
print(f"Q2 (Mediana): {Q2}")
print(f"Q3 (3º Quartil): {Q3}")
print(f"IQR (Intervalo Interquartil): {IQR}")
print(f"Whisker Inferior: {limite_inferior}")
print(f"Whisker Superior: {limite_superior}\n")
```

# 3 e 4. Outliers Acima e sua Média

```python
outliers_acima = df[df['Umidade9'] > limite_superior]['Umidade9']
qtd_outliers_acima = len(outliers_acima)
media_outliers_acima = outliers_acima.mean()

print("--- OUTLIERS ACIMA ---")
print(f"3. Quantidade de Outliers acima: {qtd_outliers_acima}")
print(f"4. Média dos Outliers acima: {media_outliers_acima}\n")
```

# 5 e 6. Outliers Abaixo e sua Média

```python
outliers_abaixo = df[df['Umidade9'] < limite_inferior]['Umidade9']
qtd_outliers_abaixo = len(outliers_abaixo)
media_outliers_abaixo = outliers_abaixo.mean()

print("--- OUTLIERS ABAIXO ---")
print(f"5. Quantidade de Outliers abaixo: {qtd_outliers_abaixo}")
print(f"6. Média dos Outliers abaixo: {media_outliers_abaixo}")
```
