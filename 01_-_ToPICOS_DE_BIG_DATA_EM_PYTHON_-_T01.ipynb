## 4 - Mostrar apenas os dias em que: - Choveu - A velocidade do vento às 9am (VelocidadeVento9) estava entre 35 e 135

```python
resultado = df[(df['ChuvaHoje'] == 'Yes') & (df['VelocidadeVento9'] >= 35) & (df['VelocidadeVento9'] <= 135)]
print(resultado)
```

## 5 - Mostrar apenas os dias que: - Não Choveu - A umidade às 9h foi maior que 80
